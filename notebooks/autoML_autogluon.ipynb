```python
import pandas as pd
import numpy as np
from autogluon.tabular import TabularPredictor, TabularDataset
from scipy import stats

class AutoMLPreprocessor:
    def __init__(self, verbosity=2):
        """
        AutoML Preprocessor for cleaning data without a target column.
        :param verbosity: Verbosity level (0=silent, 2=verbose)
        """
        self.verbosity = verbosity
        self.predictor = None  # AutoML model

    def clean_data(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Clean data using AutoGluon without requiring a target column.
        :param df: Raw DataFrame
        :return: Cleaned DataFrame
        """
        # Convert to AutoGluon format
        df = TabularDataset(df)

        # Create `temp_label`, randomly assigning 0 and 1, ensuring it is a binary classification task
        temp_label = "temp_label"
        df[temp_label] = np.random.randint(0, 2, size=len(df))

        # Initialize AutoML trainer (no real training, just for data cleaning)
        self.predictor = TabularPredictor(
            label=temp_label,
            verbosity=self.verbosity,
            problem_type="binary"
        ).fit(df, hyperparameters="default")  # Avoid unnecessary training

        # Clean the data
        cleaned_df = self.predictor.transform_features(df)

        # Remove temporary target column
        if temp_label in cleaned_df.columns:
            cleaned_df = cleaned_df.drop(columns=[temp_label])

        return cleaned_df

# Read data
file_path = 'D:\\SEGP-G6\\notebooks\\Churn_Modelling.csv'
df = pd.read_csv(file_path)

# Run AutoML cleaning
automl = AutoMLPreprocessor()
cleaned_df = automl.clean_data(df)

# Save the cleaned data
output_file_path = 'D:\\SEGP-G6\\notebooks\\output_autogluon.csv'
cleaned_df.to_csv(output_file_path, index=False)

print(f"Cleaned dataset saved to {output_file_path}")
print("Original data shape:", df.shape)
print("Cleaned data shape:", cleaned_df.shape)
```
