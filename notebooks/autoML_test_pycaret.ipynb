```python
from pycaret.classification import setup, compare_models, evaluate_model, predict_model, get_config
import pandas as pd
```

```python
file_path = 'D:\\codes\\pythoncodes\\Churn_Modelling.csv'
df = pd.read_csv(file_path)
target = 'Exited'
```

```python
automl_setup = setup(
    data=df,
    target=target,
    session_id=42,
    remove_outliers=True,
    remove_multicollinearity=True,
    normalize=True
)
```

```python
best_model = compare_models()
```

```python
print("evaluate best model...")
evaluate_model(best_model)
```

```python
test_data = get_config('X_test')
test_data_with_predictions = predict_model(best_model, data=test_data)
```

```python
test_data_with_predictions.to_csv('output.csv', index=False)
```
